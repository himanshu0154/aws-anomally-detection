```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(
    "../data/weather_cleaned.csv",
    parse_dates=['datetime']
)
```

```python
df
```

```python
df['hour'] = df['datetime'].dt.hour
df['month'] = df['datetime'].dt.month
```

```python
df['T2M_diff'] = df['T2M'].diff()
df['RH2M_diff'] = df['RH2M'].diff()
df['PS_diff'] = df['PS'].diff()
```

```python
df['T2M_roll_mean'] = df['T2M'].rolling(24).mean()
df['RH2M_roll_mean'] = df['RH2M'].rolling(24).mean()
df['PS_roll_mean'] = df['PS'].rolling(24).mean()
```

```python
df['T2M_roll_std'] = df['T2M'].rolling(24).std()
df['RH2M_roll_std'] = df['RH2M'].rolling(24).std()
df['PS_roll_std'] = df['PS'].rolling(24).std()
```

```python
df.head(30)
df.info()
df.describe()
```

```python
df['T2M_roll_dev'] = df['T2M'] - df['T2M_roll_mean']
df['RH2M_roll_dev'] = df['RH2M'] - df['RH2M_roll_mean']
df['PS_roll_dev'] = df['PS'] - df['PS_roll_mean']
```

```python
df = df.dropna().reset_index(drop=True)
```

```python
df.to_csv("../data/weather_features.csv", index=False)
```
