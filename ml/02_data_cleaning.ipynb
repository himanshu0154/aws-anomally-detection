```python
df[['T2M', 'RH2M', 'PS']] = df[['T2M', 'RH2M', 'PS']].replace(-999, np.nan)
```

```python
df.isnull().sum()
```

```python
df.dropna(subset=['T2M', 'RH2M', 'PS'], inplace=True)
```

```python
df.shape
```

```python
df['datetime'] = pd.to_datetime(
    df[['YEAR', 'MO', 'DY', 'HR']]
    .rename(columns={
        'YEAR': 'year',
        'MO': 'month',
        'DY': 'day',
        'HR': 'hour'
    })
)
```

```python
df['PS'] = df['PS']*10
```
